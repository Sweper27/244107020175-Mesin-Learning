### Deskripsi
Praktikum 1 ini bertujuan untuk memahami mekanisme kerja internal Jaringan Syaraf Tiruan (JST) 2 layer (Input Layer, 1 Hidden Layer, dan Output Layer) tanpa menggunakan pustaka tingkat tinggi (seperti Keras/PyTorch), melainkan murni menggunakan aljabar matriks NumPy.

Tahapan algoritma Backpropagation:
1. **Perambatan Maju (*Forward Pass*)**: Menghitung kombinasi linier $z = XW + b$ dan fungsi aktivasi non-linier $a = f(z)$.
2. **Perhitungan Error / Loss**: Menghitung selisih keluaran target ($y$) dengan prediksi ($a_2$) menggunakan Mean Squared Error (MSE).
3. **Perambatan Mundur (*Backward Pass*)**: Menghitung turunan parsial (gradien) dari loss terhadap bobot dan bias memanfaatkan aturan rantai (*chain rule*).
4. **Pembaruan Bobot (*Weight Update*)**: Mengoreksi nilai bobot dan bias dengan algoritma Gradient Descent: $W \leftarrow W + \eta \cdot \Delta W$.


## Langkah-Langkah Praktikum 1 (Sesuai Modul Resmi)
1. Buat dataset sederhana (XOR).
2. Inisialisasi bobot dan bias.
3. Implementasikan forward pass.
4. Hitung error dan lakukan backpropagation.
5. Update bobot menggunakan gradient descent.


```python
import numpy as np

# Set random seed untuk reprodusibilitas hasil
np.random.seed(42)

# 1. Dataset XOR (Input X berdimensi 4x2, Target y berdimensi 4x1)
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# 2. Parameter Jaringan
input_size = 2    # 2 fitur input (x1, x2)
hidden_size = 2   # 2 neuron pada hidden layer
output_size = 1   # 1 neuron output (klasifikasi biner XOR)
lr = 0.1          # Tingkat pembelajaran (learning rate)

# Inisialisasi bobot acak (Gaussian distribution) dan bias nol
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# 3. Fungsi Aktivasi Sigmoid dan Turunannya
def sigmoid(x):
    # f(x) = 1 / (1 + e^(-x))
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    # f'(x) = f(x) * (1 - f(x))
    return x * (1 - x)

# 4 & 5. Training Loop (10000 Epoch)
for epoch in range(10000):
    # --- Forward pass ---
    z1 = np.dot(X, W1) + b1       # Input ke Hidden Layer
    a1 = sigmoid(z1)              # Aktivasi Hidden Layer
    z2 = np.dot(a1, W2) + b2      # Hidden ke Output Layer
    a2 = sigmoid(z2)              # Aktivasi Output Layer (Prediksi)

    # --- Hitung error ---
    error = y - a2

    # --- Backpropagation (Backward pass) ---
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # --- Update bobot menggunakan Gradient Descent ---
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

# Output akhir
loss_final = np.mean(np.square(error))
print(f"Epoch 10000 (Selesai), Loss Akhir: {loss_final:.6f}")
print("\nPrediksi Akhir pada Dataset XOR:")
print(a2)
print("\nTarget Asli:")
print(y)
```

## Tugas 1: Eksperimen dan Komparasi Parameter JST
Sesuai instruksi modul:
1. **Ubah jumlah neuron hidden layer menjadi 3**.
2. **Bandingkan hasil loss dengan konfigurasi awal**.
3. **Tambahkan fungsi aktivasi ReLU dan bandingkan hasilnya**.


### Tugas 1.1: Mengubah Hidden Layer Menjadi 3 Neuron (Fungsi Aktivasi Sigmoid)


```python
# Eksperimen 1: Hidden Size = 3, Aktivasi = Sigmoid
np.random.seed(42)

hidden_size_exp1 = 3

# Inisialisasi bobot baru
W1_exp1 = np.random.randn(input_size, hidden_size_exp1)
b1_exp1 = np.zeros((1, hidden_size_exp1))
W2_exp1 = np.random.randn(hidden_size_exp1, output_size)
b2_exp1 = np.zeros((1, output_size))

losses_exp1 = []

for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1_exp1) + b1_exp1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2_exp1) + b2_exp1
    a2_exp1 = sigmoid(z2)

    # Hitung error
    error = y - a2_exp1

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2_exp1)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_exp1.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1_exp1 += lr * d_W1
    b1_exp1 += lr * d_b1
    W2_exp1 += lr * d_W2
    b2_exp1 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        losses_exp1.append(loss)
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

loss_final_exp1 = np.mean(np.square(error))
print(f"Epoch 10000, Loss Akhir (Hidden=3, Sigmoid): {loss_final_exp1:.6f}")
print("\nPrediksi Akhir:")
print(a2_exp1)
```

### Tugas 1.2: Menggunakan Fungsi Aktivasi ReLU pada Hidden Layer (Hidden Size = 3)
Fungsi aktivasi ReLU (*Rectified Linear Unit*) didefinisikan sebagai:
$$f(x) = \max(0, x)$$
Dan turunannya terhadap input adalah:
$$f'(x) = \begin{cases} 1 & \text{jika } x > 0 \\ 0 & \text{jika } x \le 0 \end{cases}$$


```python
# Definisi fungsi aktivasi ReLU dan turunannya
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

# Inisialisasi bobot untuk eksperimen ReLU
np.random.seed(42)
hidden_size_exp2 = 3

W1_relu = np.random.randn(input_size, hidden_size_exp2) * np.sqrt(2.0 / input_size) # He initialization
b1_relu = np.zeros((1, hidden_size_exp2))
W2_relu = np.random.randn(hidden_size_exp2, output_size) * np.sqrt(2.0 / hidden_size_exp2)
b2_relu = np.zeros((1, output_size))

lr_relu = 0.05  # Learning rate disesuaikan agar stabil pada ReLU

for epoch in range(10000):
    # Forward pass (Hidden layer memakai ReLU, Output layer tetap Sigmoid untuk probabilitas 0-1)
    z1 = np.dot(X, W1_relu) + b1_relu
    a1 = relu(z1)
    z2 = np.dot(a1, W2_relu) + b2_relu
    a2_relu = sigmoid(z2)

    # Hitung error
    error = y - a2_relu

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2_relu)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_relu.T) * relu_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1_relu += lr_relu * d_W1
    b1_relu += lr_relu * d_b1
    W2_relu += lr_relu * d_W2
    b2_relu += lr_relu * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

loss_final_relu = np.mean(np.square(error))
print(f"Epoch 10000, Loss Akhir (Hidden=3, ReLU): {loss_final_relu:.6f}")
print("\nPrediksi Akhir:")
print(a2_relu)
```

### Tugas 1.3: Tabel Komparasi dan Kesimpulan Analisis

| Konfigurasi Model | Hidden Size | Fungsi Aktivasi Hidden | Loss Awal (Epoch 0) | Loss Akhir (Epoch 10000) | Status Prediksi XOR |
| :--- | :---: | :---: | :---: | :---: | :--- |
| **Model Awal (Baseline)** | 2 | Sigmoid | 0.255830 | 0.128227 | Kurang sempurna (terjebak local minimum) |
| **Eksperimen 1** | 3 | Sigmoid | 0.318245 | **0.002297** | Berhasil memprediksi XOR dengan sangat baik |
| **Eksperimen 2** | 3 | ReLU | 0.327478 | **0.000352** | Sangat akurat & konvergensi jauh lebih cepat |

#### Kesimpulan:
1. **Pengaruh Jumlah Neuron Hidden Layer:**  
   Pada konfigurasi awal (2 neuron), kapasitas jaringan cukup terbatas sehingga model rentan terjebak pada *local minimum* (loss ~0.128, nilai prediksi pada input [0,1] dan [1,1] belum terpisah secara tegas). Ketika jumlah neuron ditingkatkan menjadi 3, jaringan memiliki derajat kebebasan dan kapasitas representasi fungsi non-linier yang lebih tinggi, sehingga loss turun drastis hingga **0.002297** dan pola XOR terpecahkan secara sempurna.

2. **Pengaruh Fungsi Aktivasi ReLU:**  
   Fungsi aktivasi ReLU mengatasi masalah *vanishing gradient* yang sering terjadi pada Sigmoid karena gradien ReLU bernilai konstan (1) untuk setiap input positif. Hal ini menyebabkan proses pembaruan bobot berlangsung jauh lebih efisien, konvergensi tercapai jauh lebih cepat (pada epoch ke-2000 loss sudah mencapai 0.003), dan menghasilkan loss akhir terendah (**0.000352**) dengan tingkat kepastian prediksi mendekati nilai 0 dan 1 mutlak.
