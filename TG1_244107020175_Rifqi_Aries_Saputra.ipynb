# **Penjelasan Library**

**PyPREP** (Python Automated PREP Pipeline)
Library khusus untuk preprocessing (pra-pemrosesan) data sinyal EEG (Elektroensefalogram / sinyal gelombang otak). PyPREP membantu mengotomatiskan pembersihan data EEG, seperti mendeteksi channel yang jelek (bad channels), melakukan referencing ulang, dan menghilangkan noise acak agar data siap dianalisis.

**Scipy** (Scientific Python)
Library fundamental untuk komputasi ilmiah dan matematika di Python. Dalam konteks pemrosesan sinyal biologis (seperti EEG atau ECG), SciPy sangat sering digunakan untuk pemrosesan sinyal lanjutan (modul scipy.signal), seperti pembuatan filter (high-pass, low-pass, notch filter), analisis frekuensi (Fast Fourier Transform / FFT), dan perhitungan statistik dasar.

**wandb** (Weights & Biases)
Platform MLOPs (Machine Learning Operations) yang digunakan untuk melacak (tracking), memvisualisasikan, dan mengorganisir eksperimen Machine Learning atau Deep Learning. Dengan wandb, Anda bisa mencatat metrics (seperti loss dan accuracy), menyimpan hiperparameter, membandingkan performa antar model secara real-time, serta membuat laporan grafik eksperimen secara otomatis.

**pyECG**
Library Python yang dirancang khusus untuk memproses dan menganalisis sinyal ECG (Elektrokardiogram / rekam detak jantung). Library ini digunakan untuk membaca format file sinyal medis (seperti ISHNE/Holter), mengekstrak peak (seperti gelombang R pada interval QRS), menganalisis variabilitas detak jantung (HRV), dan memvalidasi data sinyal jantung.
