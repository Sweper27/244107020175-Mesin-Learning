```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# 1. Load Dataset
df = pd.read_csv('wbc.csv')
print("Ukuran dataset awal:", df.shape)
```

```python
# Tugas 1
kolom_dihapus = ['id', 'Unnamed: 32']

df = df.drop(columns=[col for col in kolom_dihapus if col in df.columns])

print("Ukuran dataset setelah menghapus kolom yang tidak digunakan:", df.shape)
```

```python
# Tugas 2
# Mengubah 'M' -> 1 dan 'B' -> 0
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
# TUGAS 3
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Lakukan standarisasi pada seluruh fitur numerik (X)
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
# TAMPILKAN HASIL AKHIR
print("\n--- Output Hasil Encoding Diagnosis (y) ---")
print(y.head())

print("\n--- Output Fitur Numerik Setelah Standarisasi (X_scaled) ---")
X_scaled.head()
```
