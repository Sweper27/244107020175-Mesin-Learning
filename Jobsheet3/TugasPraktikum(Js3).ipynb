```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Membaca dataset
df = pd.read_csv('wbc.csv')
```

```python
# Membuang kolom 'id' (karena hanya identitas unik) dan 'Unnamed: 32' (karena berisi nilai NaN/kosong)
df_cleaned = df.drop(columns=['id', 'Unnamed: 32'])

# Memisahkan mana yang menjadi Fitur (X) dan mana yang menjadi Target Prediksi (y)
X = df_cleaned.drop(columns=['diagnosis']) # Variabel fitur numerik
y = df_cleaned['diagnosis']                # Variabel target

print("Kolom yang digunakan sebagai fitur:\n", X.columns.tolist())
```

```python
# Mengubah teks 'M' (Malignant) dan 'B' (Benign) menjadi angka biner
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Menampilkan perbandingan sebelum dan sesudah encoding
print("Diagnosis awal (5 data pertama):", y.head().tolist())
print("Diagnosis setelah encoding   :", y_encoded[:5])
```

```python
# Karena nilai pada fitur X rentangnya berbeda-beda (ada puluhan, ada ribuan), kita standarisasi
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Mengubah hasil yang berupa array numpy kembali menjadi DataFrame Pandas agar rapi
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Tampilan data setelah distandarisasi:")
X_scaled_df.head()
```

```python
# Membagi data dengan rasio 80% Latih (Train) dan 20% Uji (Test)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df,
    y_encoded,
    test_size=0.2,       # 0.2 mewakili 20% data uji
    stratify=y_encoded,  # Memastikan proporsi kanker M dan B seimbang di kedua pecahan data
    random_state=42      # Mengunci angka acak
)

# Melihat ringkasan jumlah datanya
print("Total keseluruhan data :", len(df_cleaned))
print("Jumlah Data Latih (80%):", len(X_train))
print("Jumlah Data Uji (20%)  :", len(X_test))
```
