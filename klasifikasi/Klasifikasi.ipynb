```python

from google.colab import files
import pandas as pd
import io

# Upload kedua dataset
print("Silakan upload reviews.csv dan cafes.csv")
uploaded = files.upload()

# Mencari nama file berdasarkan nama awal
file_reviews = next(
    (nama for nama in uploaded if nama.startswith("reviews")),
    None
)

file_cafes = next(
    (nama for nama in uploaded if nama.startswith("cafes")),
    None
)

# Memastikan kedua file ditemukan
if file_reviews is not None and file_cafes is not None:

    # Membaca dataset ulasan
    reviews = pd.read_csv(
        io.BytesIO(uploaded[file_reviews])
    )

    # Membaca dataset cafe
    cafes = pd.read_csv(
        io.BytesIO(uploaded[file_cafes])
    )

    # Menampilkan informasi dataset
    print("\nNama file ulasan:", file_reviews)
    print("Nama file cafe:", file_cafes)

    print("\nUkuran dataset reviews:", reviews.shape)
    print("Ukuran dataset cafes:", cafes.shape)

    print("\nKolom reviews:")
    print(reviews.columns.tolist())

    print("\nKolom cafes:")
    print(cafes.columns.tolist())

    print("\nContoh data reviews:")
    display(reviews.head())

    print("\nContoh data cafes:")
    display(cafes.head())

else:
    print("File belum lengkap. Pastikan kedua CSV sudah diunggah.")
```

```python
import numpy as np

# Mencari kolom berdasarkan nama yang umum digunakan.
def cari_kolom(df, kandidat):
    kolom = {c.lower().strip(): c for c in df.columns}

    for nama in kandidat:
        if nama in kolom:
            return kolom[nama]

    for c in df.columns:
        if any(nama in c.lower() for nama in kandidat):
            return c

    return None


kolom_teks = cari_kolom(
    reviews,
    ["review", "review_text", "ulasan", "komentar",
     "text", "content", "review_content"]
)

kolom_rating = cari_kolom(
    reviews,
    ["rating", "stars", "star", "score", "nilai"]
)

kolom_sentimen = cari_kolom(
    reviews,
    ["sentiment", "label", "sentimen", "polarity"]
)

print("Kolom teks:", kolom_teks)
print("Kolom rating:", kolom_rating)
print("Kolom sentimen:", kolom_sentimen)

if kolom_teks is None:
    raise ValueError(
        "Kolom teks tidak ditemukan. Periksa nama kolom pada Sel 2."
    )

data = reviews.copy()

if kolom_sentimen is not None:
    print("Kolom label ditemukan. Periksa format label sebelum melanjutkan.")
    print(data[kolom_sentimen].value_counts(dropna=False).head(20))

elif kolom_rating is not None:
    data[kolom_rating] = pd.to_numeric(
        data[kolom_rating], errors="coerce"
    )

    data = data.dropna(subset=[kolom_rating, kolom_teks])

    data["sentiment"] = np.where(
        data[kolom_rating] >= 3,
        "positif",
        "negatif"
    )

    kolom_sentimen = "sentiment"

else:
    raise ValueError(
        "Dataset tidak memiliki kolom rating atau label sentimen. "
        "Label perlu ditentukan terlebih dahulu."
    )

data = data.dropna(subset=[kolom_teks, kolom_sentimen])
data[kolom_teks] = data[kolom_teks].astype(str)
data = data[data[kolom_teks].str.strip() != ""]

print("\nDistribusi sentimen:")
print(data[kolom_sentimen].value_counts())
```

```python

import re
import pandas as pd
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Stopword bahasa Inggris dari Scikit-learn
stopwords = set(ENGLISH_STOP_WORDS)

def preprocessing(teks):
    # Case folding
    teks = str(teks).lower()

    # Menghapus URL, tanda baca, dan angka
    teks = re.sub(r"http\S+|www\.\S+", " ", teks)
    teks = re.sub(r"[^a-zA-Z\s]", " ", teks)

    # Tokenisasi dan stopword removal
    token = teks.split()
    token = [kata for kata in token if kata not in stopwords]

    return " ".join(token)

# Terapkan preprocessing tanpa stemming
data["clean_text"] = data[kolom_teks].map(preprocessing)

# Hapus teks kosong dan ulasan duplikat
data = data[data["clean_text"].str.strip() != ""].copy()
data = data.drop_duplicates(subset=["clean_text"])

display(data[[kolom_teks, "clean_text", kolom_sentimen]].head())
print("Jumlah data setelah preprocessing:", len(data))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

X_train_text, X_test_text, y_train, y_test = train_test_split(
    data["clean_text"],
    data[kolom_sentimen],
    test_size=0.20,
    random_state=42,
    stratify=data[kolom_sentimen]
)

# Bag of Words: mengubah teks menjadi hitungan fitur kata
vectorizer = CountVectorizer()

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print("Jumlah data latih:", X_train.shape[0])
print("Jumlah data uji:", X_test.shape[0])
print("Jumlah fitur kata:", X_train.shape[1])
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import (
    GaussianNB,
    MultinomialNB
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

models = {
    "KNN": KNeighborsClassifier(n_neighbors=5),

    "SVM / SVC": SVC(kernel="linear"),

    "Gaussian Naive Bayes": GaussianNB(),

    "Multinomial Naive Bayes": MultinomialNB()
}

# GaussianNB memerlukan array dense
X_train_dense = X_train.toarray()
X_test_dense = X_test.toarray()

hasil = []
model_terlatih = {}

for nama, model in models.items():
    if nama == "Gaussian Naive Bayes":
        model.fit(X_train_dense, y_train)
        prediksi = model.predict(X_test_dense)
    else:
        model.fit(X_train, y_train)
        prediksi = model.predict(X_test)

    model_terlatih[nama] = model

    hasil.append({
        "Algoritma": nama,
        "Accuracy": accuracy_score(y_test, prediksi),
        "Precision": precision_score(
            y_test, prediksi,
            average="binary" if len(set(y_test)) == 2 else "weighted",
            pos_label="positif",
            zero_division=0
        ) if set(y_test).issubset({"positif", "negatif"}) else
        precision_score(y_test, prediksi, average="weighted", zero_division=0),
        "Recall": recall_score(
            y_test, prediksi,
            average="binary" if len(set(y_test)) == 2 else "weighted",
            pos_label="positif",
            zero_division=0
        ) if set(y_test).issubset({"positif", "negatif"}) else
        recall_score(y_test, prediksi, average="weighted", zero_division=0),
        "F1-score": f1_score(
            y_test, prediksi,
            average="weighted",
            zero_division=0
        )
    })

tabel_hasil = pd.DataFrame(hasil)
display(
    tabel_hasil.sort_values(
        "Accuracy", ascending=False
    ).reset_index(drop=True)
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

for nama, model in model_terlatih.items():
    if nama == "Gaussian Naive Bayes":
        prediksi = model.predict(X_test_dense)
    else:
        prediksi = model.predict(X_test)

    print("\n" + "=" * 50)
    print(nama)
    print(classification_report(
        y_test, prediksi, zero_division=0
    ))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        prediksi,
        xticks_rotation="vertical"
    )

    plt.title(f"Confusion Matrix - {nama}")
    plt.tight_layout()
    plt.show()
```
